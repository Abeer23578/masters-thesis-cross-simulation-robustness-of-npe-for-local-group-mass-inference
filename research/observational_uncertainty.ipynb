# Observational uncertainty

The six headline models remain noise-free because their estimand is the simulation-catalogue conditional distribution. Real Local Group inference requires an additional observation model. Do not use measurement noise to align FLAMINGO and TNG.

Shared IMF/population-synthesis assumptions correlate stellar-mass systematics. M31 mass and tangential velocity depend on distance. Galactocentric transformations introduce further correlations. The catalogue's all-bound stellar masses and observed integrated stellar masses also require a definition/IMF comparison. The old halo-mass-ratio proxy is excluded from the new application; it imports external halo-mass information.

## Probabilistic target

For a selected catalogue population S, latent exact observables x and measured y, assume y is conditionally independent of theta given x and the observation model is K(y|x). Then

p_S(theta|y) = integral p_S(theta|x) K(y|x) p_S(x) dx / integral K(y|x) p_S(x) dx.

Thus averaging q(theta|x) over x sampled from r(x|y) is valid only if r is the latent-observable posterior under the SAME population model, or if importance weights K(y|x)p_S(x)/r(x|y) are applied. A Gaussian centred on y is generally a normalized likelihood under a flat observable prior; it need not equal p_S(x|y). An external measurement posterior already contains a prior and must be adjusted for that prior before combination. Clipping to the selected catalogue changes the population question; selection boundaries must be disclosed.

This identity is a direct Bayes-rule derivation. It corrects the interpretation of the historical perturb-and-average calculation without overwriting it: those saved outputs are uncertainty-sensitivity mixtures under an adopted input distribution, not demonstrated exact measurement-marginalized posteriors.

## Methods considered and decision

* Noise-aware training: sample y from a documented K(y|x) for training rows, then fit q(theta|y). It integrates latent observables under the empirical selected population. Split hosts first; fit scalers on training rows; validate with independent noise draws. Use a common physical observation model and transform to each simulation's velocity convention. A small separate dynamics-only application is the preferred future extension once K is established.
* Inference-time integration: valid with the weighting identity above. A fitted feature density introduces an additional density-estimation assumption; finite-catalogue likelihood weighting is an empirical alternative whose effective sample size can collapse at precise observations. Require effective sample size and support diagnostics.
* Explicit uncertainty inputs: appropriate for varying known measurement precision; train over the same precision distribution and condition on it. Unnecessary for one fixed Local Group likelihood and does not supply missing covariance.
* Naive Monte Carlo perturbation: retain only as labelled sensitivity, not the primary posterior.
* A neural likelihood/joint model can integrate measurement noise explicitly but would expand the thesis beyond the smallest final experiment.

[Deistler et al., Simulation-Based Inference: A Practical Guide, arXiv:2508.12939](https://arxiv.org/html/2508.12939v1) includes measurement noise in the forward model. [Cranmer, Brehmer & Louppe (2020), arXiv:1911.01429](https://arxiv.org/abs/1911.01429), DOI 10.1073/pnas.1912789117, reviews forward-model SBI. These support the observation-model approach; the integration identity above states the assumptions needed here.

No final uncertainty-aware Local Group mass is reported now. Required inputs are a reproducible joint distance/proper-motion/radial/solar-motion likelihood (or samples with their prior identified), and, for stellar conditioning, consistent stellar-mass definitions and systematics. The observational notebook records the available sources and validates the integration algebra on a synthetic model. It does not invent missing covariance or claim that the actual observation model has been validated.

Posterior scatter conditional on a simulation, observational error, differences across training domains, component-bootstrap uncertainty and finder-definition systematics remain separate quantities. A systematic Delta shift belongs to the mapping diagnostic, not K(y|x).
