# Model comparison

The six models compare three training populations—FLAMINGO, TNG300 and their equal-domain combination—with two feature sets. Dynamics uses separation, peculiar radial velocity and tangential speed. Stellar sum adds `log10(Mstar,a + Mstar,b)`.

The target is the common host's M200c for same-host pairs and the sum of the two host M200c values for different-host pairs. Stellar masses are all-bound catalogue values. The [11.5,14.5] box limits posterior sampling; the empirical catalogue mass distribution supplies the effective training prior.

## Splits and preprocessing

Host-connected components stay together across train, validation and test. FLAMINGO has 109150/12154/21409 rows and TNG300 has 1169/130/231. The TNG test set contains 183 same-host and 48 different-host pairs. The test populations were also inspected in earlier analyses, so they are reused test sets rather than prospective validation.

Feature scaling uses only training rows. The combined scaler gives the two simulations equal total weight. Validation is component-disjoint and the best validation checkpoint is restored.

## Training

The scalar MAF has five transforms, 64 hidden features, batch size 64, learning rate 0.00016488693479340492, a maximum of 200 epochs and patience 20. The Optuna `num_layers` parameter was not connected to the estimator and is not activated here.

Combined training minimizes `0.5 mean_F[-log q(theta|x)] + 0.5 mean_T[-log q(theta|x)]`. Each batch samples 32 rows from each simulation with replacement. Early stopping uses the equally weighted domain validation losses. Reusing TNG rows changes their exposure, not the number of independent systems.

The scalar affine flow is Gaussian in log mass before support truncation. Equal-domain training therefore does not imply an exact 50/50 mixture of the two specialist posteriors at every input.

## Evaluation and sensitivity tests

The held-out comparison uses 500 posterior samples per pair. Accuracy, signed bias, rank diagnostics, credible-interval coverage and widths are reported together. Paired intervals use 2000 connected-component resamples on the same evaluation rows, conditional on the fitted models and saved draws.

Stellar normalization uses separate sum/individual checkpoints. Its shift is calibrated on training rows from a forward bound-mass relation, independently of held-out M200c errors. The same 231 TNG test rows are evaluated at every shift. The observed Local Group comparison separately reports nominal inputs and an approximate mixture over observational uncertainty.
