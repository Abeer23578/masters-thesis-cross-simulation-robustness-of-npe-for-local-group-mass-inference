# Cross-simulation robustness of Local Group mass inference

This thesis investigates the robustness of NPE under cross-simulation transfer of FLAMINGO and TNG300.


```python
from pathlib import Path
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display, Markdown

ROOT = next(p for p in [Path.cwd().resolve(), *Path.cwd().resolve().parents]
            if (p / 'pyproject.toml').is_file())
def table(relative):
    return pd.read_csv(ROOT / 'results' / relative)
def saved_json(relative):
    return json.loads((ROOT / 'results' / relative).read_text(encoding='utf-8'))
plt.rcParams.update({'figure.figsize': (9, 4), 'font.size': 10,
                     'axes.spines.top': False, 'axes.spines.right': False})
pd.set_option('display.max_rows', 40)
pd.set_option('display.max_columns', 16)
```

Large-scale hydrodynamical simulations such as IllustrisTNG and FLAMINGO simulate
billions of particles across cosmic time by incorporating sophisticated physical models.
However, different simulations vary in numerical methods, calibration strategies, and
subgrid physics. This makes inferences highly simulation-dependent.

The main question is which individual observables transfer well across simulations and
which observables drive the generalization breakdown when used for training. The
notebooks follow this question from catalogue construction to the observed Local Group.

### Notebook order

Notebooks 01–03 cover the data, exploratory analysis and methods. Notebooks 04–07
compare prediction, calibration and bootstrap references. Notebooks 08–11 test competing
explanations, and 12 applies the models to the observed Local Group. Notebook 13 explores
classifier-weighted mixtures of the specialist posteriors.

The [thesis map](../docs/thesis_map.ipynb) locates each table and figure.
[Scientific notes](../docs/scientific_notes.ipynb) give further definitions and interpretation.

### Three different mixtures

A **combined NPE** is trained on balanced simulation minibatches. An **observational-input
mixture** averages conditional posteriors over sampled measurements. A **classifier-weighted
specialist mixture** combines the two separately trained simulation posteriors using a
feature-based diagnostic weight.
