## Interpretive limits

The fixed-stellar-window mechanism has the expected sign if TNG stellar masses are lower at fixed halo mass. No numerical dilution follows from comparing0.244 to0.30–0.36 dex: slopes, scatter, halo abundances, host composition and pair selection matter, and the quoted offsets condition on different quantities. A halo-mass-matched selection creates mass agreement by construction; investigate selection probabilities and the remaining stellar relation instead of treating mass matching itself as confirmation.

The disc-fraction contrast does not establish stronger disc suppression or stronger feedback in FLAMINGO. Its fiducial AGN model is thermal, not kinetic. The global second moment includes rotation; kappa measures a fraction of kinetic energy, so a similar total second moment and a different rotational fraction are not contradictory.

Avoid stating that the real Local Group has one globally measured solar metallicity comparable to all-bound mass-weighted simulation metallicities. Describe the earlier adopted near-solar reference and the measurement mismatch; no new observational metallicity audit was performed here.

The m9 claim that it excludes71–78% of the stellar window is not verified.100*initial_mgas=1.07e11 solar masses is only a proxy, not the actual100-star-particle criterion, and fractions depend on which population is counted. Do not use the proxy fractions output by the script as actual m9 loss fractions.
