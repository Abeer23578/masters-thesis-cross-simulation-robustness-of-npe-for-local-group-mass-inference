# From the thesis to the notebooks

This repository follows *Cross-Simulation Robustness of Neural Posterior Estimation for Local Group Mass Inference* (). The main notebooks follow the scientific argument from catalogue selection to the observed Local Group. Stellar normalization, common support, bootstrap experiments, and the alternative mass target are part of that main notebook sequence.

The map covers the 32 numbered tables and 16 numbered figures in the thesis. It identifies the experiment, the saved evidence, and the earlier workflow used to produce it. It does not claim that each original visual has been rebuilt with exactly the same layout. Some notebooks display the saved numerical results; others regenerate a plot from those results. Raw catalogue reconstruction and model training are separate, optional reproduction steps.

## Experiments that do not have their own numbered figure or table

- The sample-size posterior reference in [06](../notebooks/06_bootstrap_and_sample_size.ipynb) uses the earlier [bootstrap workflow](../research/notebooks/original/npe_bootstrap.ipynb), [posterior-subsampling repeats](../results/bootstrap/posterior_subsampling_repeats.csv), [training-noise repeats](../results/bootstrap/training_noise_repeats.csv), [noise comparison](../results/bootstrap/noise_comparison.csv), and [observed JSD](../results/bootstrap/observed_jsd.json). These are readable exports of the established pickle results. A 200-refit reference is not the same as resampling saved test predictions. [Ablation repeat results](../results/bootstrap/ablation_repeats.csv) retain the related feature-specific references.
- The prior sensitivity discussed in §6.2 is presented in [11](../notebooks/11_prior_and_support_sensitivity.ipynb), using training-only KDEs, [density-estimation details](../results/prior/density_estimation_details.csv), [held-out prior swaps](../results/prior/prior_swap_test_metrics.csv), and [Local Group prior swaps](../results/prior/local_group_prior_swap.csv). Bandwidth variation and effective sample size are part of the result.
- The box-bound sensitivity in [11](../notebooks/11_prior_and_support_sensitivity.ipynb) is a supplementary check, with [held-out](../results/bounds/heldout_bound_sensitivity.csv) and [Local Group](../results/bounds/local_group_bound_sensitivity.csv) results. It changes posterior support, not the empirical training prior.
- The earlier MAF/NSF/MDN comparison is included in [research](../research/notebooks/original/npe_model_family_comparison.ipynb) and `results/model_family`. It does not establish the behaviour of final stellar-sum or combined models under those alternative estimators.
- The classifier-weighted specialist mixtures in [13](../notebooks/13_classifier_mixture_posteriors.ipynb) are an extension. They do not replace the two main thesis experiments or establish a preferred simulation.
