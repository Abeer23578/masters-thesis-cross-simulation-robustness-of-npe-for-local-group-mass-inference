# Reproducing the analyses

All 14 result notebooks use the public aggregate tables in `results/`. Install `requirements.txt` and run them from top to bottom. Reanalysing individual posterior draws requires the saved NPZ files; new inference requires the matching model checkpoints and environment.

The local folder includes available caches and checkpoints, which are excluded from Git. A public clone contains the summary tables. There is no public release of the derived artifact bundle yet. [Notebook 90](../notebooks/90_generate_catalogue_caches.ipynb) generates catalogue caches; obtaining the exact fitted networks and original Monte Carlo draws requires their saved artifacts.

## Import saved artifacts

```bash
python scripts/import_artifacts.py --source /path/to/catalogue_project --source-label original_project
python scripts/import_artifacts.py --source /path/to/saved_models --source-label final_archive
```

`provenance/artifacts.json` lists the required files and checksums. The importer checks file content, copies missing artifacts and stops if an existing destination differs. It does not deserialize pickle files. Additional diagnostic artifacts use the source label `thesis_workspace`.

The six-model registry is `research/out/final_six/model_registry_v1.json`. Saved evaluations are grouped under `research/out/final_six`, stellar-shift results under `research/out/final_delta`, and observed Local Group results under `research/out/local_group_final`.

## Reanalyse or refit

[Notebook 92](../notebooks/92_reproduce_saved_analyses.ipynb) gives the command order for empirical-prior swaps, support-bound sensitivity and classifier diagnostics. These analyses reuse the existing models and samples.

Detailed workflows are in `research/notebooks/original/`. Set `LG_RUN_RESEARCH_WORKFLOW=1` to enable their calculation cells, `LG_PROJECT` to the `research/` folder and `TNG_BASE` to the local TNG catalogue. Use the environment in `requirements-research.txt`; older block-grid models also depend on packages recorded in `research/env_roihu.txt`, including Optuna, ArviZ and PyTorch.

Refitting writes new outputs. It need not reproduce the exact original Monte Carlo draws. The stellar normalization experiment uses the separate mass-comparison checkpoints, not the six models used for the final Local Group estimates.

The detailed refit workflows also require the archived split and calibration bundle, with the relevant reference checkpoints. Generating catalogue caches in notebook 90 alone does not supply those artifacts. Raw catalogue extraction and new NPE training have not been independently rerun for this repository.

## Scope

The saved final comparisons include the main six-model, stellar-normalization and Local Group results. Two older zero-shift archives were unavailable; the completed stellar-shift comparison has its own zero-shift draws and paired intervals.

Host-conditioned training, new density families for the final stellar models, seed ensembles, controlled feedback/finder experiments and new parent-catalogue selections remain follow-up work.

The research regression suite uses synthetic fixtures. Its archive-integrity test runs when the archived run bundle is present and skips when that optional bundle is absent.
