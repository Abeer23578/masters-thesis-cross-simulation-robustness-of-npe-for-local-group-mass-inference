# Publishing

Create an empty GitHub repository, then initialize this folder and add the remote. Use a verified or GitHub noreply address for commits.

```bash
git init
git config user.name "Abeer Ayyaz"
git config user.email "YOUR_GITHUB_VERIFIED_OR_NOREPLY_EMAIL"
git add .
git status --short
git diff --cached --stat
git commit -m "Add thesis notebooks and analysis code"
git branch -M main
git remote add origin YOUR_REPOSITORY_URL
git push -u origin main
```

Check the staged files before committing. `.gitignore` excludes pickle files, raw catalogue arrays, model checkpoints, posterior archives and local outputs. The small result tables stay in the repository so the analysis notebooks can run without those files.

A derived-data release can be linked here once available. Include its checksum and follow the simulation data terms when distributing catalogues. Choose a software license before publishing.
