# Start here

Use Python 3.11 or newer. From this folder on Windows:

```powershell
py -3.11 -m venv .venv
.\.venv\Scripts\python -m pip install -r requirements.txt
.\.venv\Scripts\python -m jupyter lab
```

On Linux or macOS:

```bash
python3 -m venv .venv
.venv/bin/python -m pip install -r requirements.txt
.venv/bin/python -m jupyter lab
```

Open `notebooks/00_start_here.ipynb`, then follow notebooks 01-13 in order.
Each reading notebook can also run independently. They load committed aggregate
CSV/JSON files and use CPU computations; neither Roihu nor a GPU is required.

For raw catalogue generation, install `requirements-data.txt` in a separate
environment. TNG API access uses `TNG_API_KEY` in your environment. Follow
`notebooks/90_generate_catalogue_caches.ipynb` and the data guide.
`requirements-research.txt` records the original Roihu package pins, including
versions needed by historical pickles. It is provenance for model reuse,
not an extra requirement for reading the results. Torch was supplied by the
original CSC module, so consult the reproduction guide before model loading.

Run `python scripts/check_repository.py` to validate notebook structure,
links, source syntax and the public file inventory. Add `--execute` to run
the 14 result notebooks and save executed copies under ignored `outputs/`.
